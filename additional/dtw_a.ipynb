# DTW - Dynamic Time Warping (overview)

```python
from dtw import dtw
import numpy as np

series_a = np.array([1.0, 1.2, 1.5, 1.3, 1.8, 2.0, 1.7, 1.6, 1.5, 1.0])
series_b = np.array([0.8, 1.0, 1.2, 0.9, 1.5, 1.8, 2.0, 1.6, 1.2])

alignment = dtw(series_a, series_b, keep_internals=True)

print("Odległość DTW:", alignment.distance)

alignment.plot(type="twoway")
```

Algorithm found the best alignment for the distance 1.39999
